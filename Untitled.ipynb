```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set seed for reproducibility
np.random.seed(123)

# Parameters
n_subjects = 100
target_index = 41  # Subject 42 (index starts at 0)
fixed_subjective = 6.0
fixed_objective = 0.4

# Subject IDs
subject_ids = [f"Subject_{i+1}" for i in range(n_subjects)]

# Prepare simulations
simulations = []

for sim_id in range(1, 4):
    if sim_id == 1:
        # Simulation 1: Normal distributions
        subjective = np.random.normal(loc=4, scale=1.5, size=n_subjects).clip(0, 8)
        objective = np.random.normal(loc=0.5, scale=0.15, size=n_subjects).clip(0, 1)
    elif sim_id == 2:
        # Simulation 2: Skewed subjective, uniform objective
        subjective = (np.random.beta(a=2, b=5, size=n_subjects) * 8).clip(0, 8)
        objective = np.random.uniform(0, 1, size=n_subjects)
    elif sim_id == 3:
        # Simulation 3: Bimodal subjective, normal objective
        subjective = np.concatenate([
            np.random.normal(loc=2, scale=0.5, size=n_subjects//2),
            np.random.normal(loc=6, scale=0.5, size=n_subjects//2)
        ])
        np.random.shuffle(subjective)
        subjective = np.clip(subjective, 0, 8)
        objective = np.random.normal(loc=0.5, scale=0.2, size=n_subjects).clip(0, 1)

    # Set fixed scores for Subject 42
    subjective[target_index] = fixed_subjective
    objective[target_index] = fixed_objective

    # Compute z-scores
    subj_z = (subjective - np.mean(subjective)) / np.std(subjective)
    obj_z = (objective - np.mean(objective)) / np.std(objective)
    awareness_z = subj_z - obj_z

    # Create DataFrame
    df = pd.DataFrame({
        'Subject': subject_ids,
        'Simulation': f"Sim_{sim_id}",
        'Subjective': subjective,
        'Objective': objective,
        'Subjective_z': subj_z,
        'Objective_z': obj_z,
        'Awareness_zDiff': awareness_z
    })
    simulations.append(df)

# Combine all simulations
df_all = pd.concat(simulations, ignore_index=True)
df_target = df_all[df_all['Subject'] == 'Subject_42']

# --- Plot awareness score across simulations for Subject 42 ---
plt.figure(figsize=(10, 5))
sns.barplot(
    data=df_target,
    x="Simulation",
    y="Awareness_zDiff",
    palette="coolwarm"
)
plt.axhline(0, color='gray', linestyle='--')
plt.title("Awareness Score (z-subjective − z-objective) for Subject 42\nAcross Different Group Distributions")
plt.ylabel("Awareness Score (z-difference)")
plt.xlabel("Simulation")
plt.tight_layout()
plt.show()

# --- Plot distribution of subjective/objective scores with Subject 42 highlighted ---
fig, axes = plt.subplots(3, 2, figsize=(12, 10), sharex='col')

for i, sim in enumerate(['Sim_1', 'Sim_2', 'Sim_3']):
    sim_data = df_all[df_all['Simulation'] == sim]

    sns.histplot(sim_data['Subjective'], kde=True, ax=axes[i, 0], color='skyblue')
    axes[i, 0].set_title(f"{sim} - Subjective Scores")
    axes[i, 0].axvline(fixed_subjective, color='red', linestyle='--', label='Subject 42')
    axes[i, 0].legend()

    sns.histplot(sim_data['Objective'], kde=True, ax=axes[i, 1], color='orange')
    axes[i, 1].set_title(f"{sim} - Objective Scores")
    axes[i, 1].axvline(fixed_objective, color='red', linestyle='--', label='Subject 42')
    axes[i, 1].legend()

plt.tight_layout()
plt.show()
```
